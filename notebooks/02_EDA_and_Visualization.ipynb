# 📊 Notebook 02: Exploratory Data Analysis
## Network Intrusion Detection System — NSL-KDD

**Objective:** Understand the data through visualizations  
**Plots generated:**
- Class distribution (bar + pie)
- Feature correlation heatmap
- Feature distributions
- Box plots by class

```python
# ─── Imports ────────────────────────────────────────────────────
import sys
import os
import warnings
warnings.filterwarnings('ignore')

PROJECT_ROOT = os.path.abspath(os.path.join(os.getcwd(), '..'))
sys.path.insert(0, PROJECT_ROOT)

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from src.utils import load_numpy, load_model, print_header
from src.visualization import (
    set_plot_style, plot_class_distribution, plot_pie_chart,
    plot_correlation_heatmap, plot_feature_distributions, plot_boxplots
)
from src.feature_engineering import get_correlation_matrix

set_plot_style()
print("✅ All imports successful")
```

```python
# ─── Load Processed Data ────────────────────────────────────────
processed_dir = os.path.join(PROJECT_ROOT, 'data', 'processed')

X_train = load_numpy(os.path.join(processed_dir, 'X_train.npy'))
X_test = load_numpy(os.path.join(processed_dir, 'X_test.npy'))
y_train_binary = load_numpy(os.path.join(processed_dir, 'y_train_binary.npy'))
y_test_binary = load_numpy(os.path.join(processed_dir, 'y_test_binary.npy'))
y_train_multi = load_numpy(os.path.join(processed_dir, 'y_train_multi.npy'))
y_test_multi = load_numpy(os.path.join(processed_dir, 'y_test_multi.npy'))
feature_names = load_numpy(os.path.join(processed_dir, 'feature_names.npy'))

print(f"\nX_train: {X_train.shape}")
print(f"y_train_binary: {y_train_binary.shape}")
print(f"y_train_multi: {y_train_multi.shape}")
print(f"Features: {len(feature_names)}")
```

```python
# ─── 1. Binary Class Distribution ──────────────────────────────
print_header("Binary Class Distribution")

plot_class_distribution(
    y_train_binary,
    "Training Set — Binary Distribution (0=Normal, 1=Attack)",
    save_path=os.path.join(PROJECT_ROOT, 'plots', 'eda',
                           'class_distribution_binary.png')
)

plot_pie_chart(
    y_train_binary,
    {0: 'Normal', 1: 'Attack'},
    "Binary Class Distribution",
    save_path=os.path.join(PROJECT_ROOT, 'plots', 'eda',
                           'pie_chart_binary.png')
)
```

```python
# ─── 2. Multiclass Distribution ────────────────────────────────
print_header("Multiclass Distribution")

# Load label encoder to get class names
le_multi = load_model(os.path.join(PROJECT_ROOT, 'models', 'encoders',
                                    'label_encoder_multi.pkl'))
class_names = dict(enumerate(le_multi.classes_))
print(f"Class mapping: {class_names}")

plot_class_distribution(
    y_train_multi,
    "Training Set — Multiclass Distribution",
    save_path=os.path.join(PROJECT_ROOT, 'plots', 'eda',
                           'class_distribution_multi.png'),
    multi=True
)

plot_pie_chart(
    y_train_multi,
    class_names,
    "Multiclass Attack Distribution",
    save_path=os.path.join(PROJECT_ROOT, 'plots', 'eda',
                           'pie_chart_multi.png')
)
```

```python
# ─── 3. Correlation Heatmap ────────────────────────────────────
print_header("Feature Correlation Analysis")

# Use only numerical features for correlation (first 38 features)
# to keep the heatmap readable
num_numerical = 38  # Original numerical features before encoding
X_train_num = X_train[:, :num_numerical]
feat_names_num = feature_names[:num_numerical]

corr_matrix, high_corr = get_correlation_matrix(
    X_train_num, feat_names_num, threshold=0.90)

print(f"\n🔗 Highly correlated pairs (|r| ≥ 0.90):")
for f1, f2, r in high_corr[:15]:
    print(f"   {f1:35s} ↔ {f2:35s}  r = {r:.4f}")

plot_correlation_heatmap(
    corr_matrix,
    "Feature Correlation Heatmap (Numerical Features)",
    save_path=os.path.join(PROJECT_ROOT, 'plots', 'eda',
                           'feature_correlation_heatmap.png')
)
```

```python
# ─── 4. Feature Distributions ──────────────────────────────────
print_header("Feature Distributions")

plot_feature_distributions(
    X_train, feature_names, y_train_binary, top_n=10,
    save_path=os.path.join(PROJECT_ROOT, 'plots', 'eda',
                           'feature_distributions.png')
)
```

```python
# ─── 5. Box Plots ──────────────────────────────────────────────
print_header("Box Plots by Class")

plot_boxplots(
    X_train, feature_names, y_train_binary, top_n=10,
    save_path=os.path.join(PROJECT_ROOT, 'plots', 'eda',
                           'feature_boxplots.png')
)
```

## 📝 Key Insights

1. **Class Imbalance:** Normal traffic significantly outnumbers attacks
   in the binary case. In multiclass, U2R and R2L are extremely rare.

2. **Correlated Features:** Several dst_host_* features are highly
   correlated — candidates for dimensionality reduction (Module 6).

3. **Discriminative Features:** Features like `src_bytes`, `dst_bytes`,
   `count`, and `srv_count` show clear separation between normal
   and attack traffic.

4. **Outliers:** Box plots reveal significant outliers in byte-count
   features, which is expected for DoS attacks.
